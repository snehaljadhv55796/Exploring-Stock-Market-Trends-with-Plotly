1. Import Libraries

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

2. Load Dataset

```python
df = pd.read_csv("/content/Stock_data.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

3. Data Cleaning
Check missing values

```python
df.isnull().sum()
```

Check duplicates

```python
df.duplicated().sum()
```

Convert Date

```python
df["Date"] = pd.to_datetime(df["Date"])
```

Sort by Date

```python
df = df.sort_values("Date").reset_index(drop=True)
df.head()
```

4. Stock Price Trend

```python
fig = px.line(
    df,
    x="Date",
    y="Close",
    title="Stock Closing Price Trend",
    labels={
        "Date": "Date",
        "Close": "Closing Price"
    }
)

fig.show()
```

Insight

This chart helps identify periods where the stock price increased or decreased.

5. Open vs Close Price

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["Open"],
        mode="lines",
        name="Open Price"
    )
)

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["Close"],
        mode="lines",
        name="Close Price"
    )
)

fig.update_layout(
    title="Opening Price vs Closing Price",
    xaxis_title="Date",
    yaxis_title="Price"
)

fig.show()
```

6. Highest and Lowest Price

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["High"],
        mode="lines",
        name="High"
    )
)

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["Low"],
        mode="lines",
        name="Low"
    )
)

fig.update_layout(
    title="Daily High and Low Stock Prices",
    xaxis_title="Date",
    yaxis_title="Price"
)

fig.show()
```

7. Candlestick Chart

```python
fig = go.Figure(
    data=[
        go.Candlestick(
            x=df["Date"],
            open=df["Open"],
            high=df["High"],
            low=df["Low"],
            close=df["Close"],
            name="Stock"
        )
    ]
)

fig.update_layout(
    title="Stock Price Candlestick Chart",
    xaxis_title="Date",
    yaxis_title="Price",
    xaxis_rangeslider_visible=False
)

fig.show()
```

9. Total Trade Quantity

```python
fig = px.line(
    df,
    x="Date",
    y="Total Trade Quantity",
    title="Total Trade Quantity Over Time"
)

fig.show()
```

```python
fig = px.bar(
    df,
    x="Date",
    y="Total Trade Quantity",
    title="Daily Trading Quantity"
)

fig.show()
```

11. Trading Quantity vs Turnover

```python
fig = px.scatter(
    df,
    x="Total Trade Quantity",
    y="Turnover (Lacs)",
    title="Trade Quantity vs Turnover",
    hover_data=["Date", "Close"]
)

fig.show()
```

12. Daily Price Change

```python
df["Price_Change"] = df["Close"].diff()
```

```python
df[["Date", "Close", "Price_Change"]].head()
```

13. Daily Return

```python
df["Daily_Return"] = df["Close"].pct_change() * 100
```

```python
fig = px.line(
    df,
    x="Date",
    y="Daily_Return",
    title="Daily Stock Returns"
)

fig.show()
```

14. Moving Average

20-Day Moving Average

```python
df["MA20"] = df["Close"].rolling(20).mean()
```

50-Day Moving Average

```python
df["MA50"] = df["Close"].rolling(50).mean()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["Close"],
        mode="lines",
        name="Close Price"
    )
)

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["MA20"],
        mode="lines",
        name="20-Day MA"
    )
)

fig.add_trace(
    go.Scatter(
        x=df["Date"],
        y=df["MA50"],
        mode="lines",
        name="50-Day MA"
    )
)

fig.update_layout(
    title="Stock Price with Moving Averages",
    xaxis_title="Date",
    yaxis_title="Price"
)

fig.show()
```

15. Volatility Analysis

```python
df["Volatility"] = df["Daily_Return"].rolling(20).std()
```

```python
fig = px.line(
    df,
    x="Date",
    y="Volatility",
    title="20-Day Rolling Volatility"
)

fig.show()
```

16. Monthly Analysis

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)
```

```python
monthly_close = df.groupby("Month")["Close"].mean().reset_index()
```

```python
fig = px.line(
    monthly_close,
    x="Month",
    y="Close",
    title="Monthly Average Closing Price"
)

fig.update_xaxes(tickangle=45)

fig.show()
```

17. Monthly Turnover

```python
monthly_turnover = (
    df.groupby("Month")["Turnover (Lacs)"]
    .sum()
    .reset_index()
)
```

```python
fig = px.bar(
    monthly_turnover,
    x="Month",
    y="Turnover (Lacs)",
    title="Monthly Turnover"
)

fig.update_xaxes(tickangle=45)

fig.show()
```
